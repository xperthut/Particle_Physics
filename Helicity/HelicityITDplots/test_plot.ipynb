```python
import Plot
import numpy as np
%matplotlib inline
import matplotlib as mpl
import matplotlib.patches as mpatches

mpl.rcParams['mathtext.fontset'] = 'cm'
mpl.rcParams['mathtext.rm'] = 'serif'
```

```python
def loadData(filename):
    with open(filename, 'r') as File:
        Data = [tuple(float(x) for x in line.split()) for line in File
                if not line.startswith("#")]
        return tuple(zip(*Data))
    
def plot_data(ifn, ft='XGB', fc='gray'):
    # Set some style-related things.
    Plot.Plot.init()

    # Create a plot object with figure size.
    Plotter = Plot.Plot((5, 4)); # If no argument, default is (4, 3)
    Plotter.newPlot();
    Plotter.setXRange(0.0, 15.01);
    Plotter.setYRange(0.0, 0.121);

    Plotter.Plot.set_ylabel(r'$\mathit{\Delta \mathfrak{M}_g(\omega)}$', fontsize =14)
    Plotter.Plot.set_xlabel(r'$\mathit{\omega}$', fontsize =14)
    #Plotter.setXLabel(r"$t_1 - t_2 /2$")
    Plotter.Plot.xaxis.set_label_coords(0.5, -0.08)
    Plotter.Plot.yaxis.set_label_coords(-0.12, 0.5)
    Plotter.Plot.yaxis.set_ticks(np.arange(0.0, 0.121, 0.02))
    Plotter.Plot.yaxis.set_ticklabels([ 0.0, 0.020, 0.040, 0.060, 0.080, 0.10, 0.120], fontsize=10)

    Plotter.Plot.xaxis.set_ticks(np.arange(0.0, 15.01, 2))
    Plotter.Plot.xaxis.set_ticklabels([0, 2, 4, 6, 8, 10, 12, 14], fontsize=10)

    Plotter.Plot.fill_between(*(loadData(f"{ifn}_{ft}_band.txt")), facecolor=fc, alpha=0.55, linewidth=0)
    p4 = Plotter.Plot.fill(0,0,fc, alpha=0.55, linewidth=0)
    p44 = Plotter.plotLine(loadData(f"{ifn}_{ft}_line.txt"), color=fc, linewidth=1.5, alpha=1.0, label=" legend");
    Plotter.Plot.legend([(p4[-1], p44[0])], [ r' $\mathit{\Delta \mathfrak{M}_g(\omega)}$ '+f'({ft})'], numpoints=1,prop={'size':12.0},loc= (0.55, 0.8))
    Plotter.save(f"DeltaMg-{ft}_v1.pdf");
    
```

# RF

```python
plot_data(ifn='pITD', ft='RF', fc='c')
```
